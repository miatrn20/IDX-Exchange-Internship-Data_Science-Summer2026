```python
import joblib

metadata = joblib.load("model_metadata.pkl")

numeric_features = metadata["numeric_features"]
categorical_features = metadata["categorical_features"]

print("NUMERIC FEATURES")
for feature in numeric_features:
    print("-", feature)

print("\nCATEGORICAL FEATURES")
for feature in categorical_features:
    print("-", feature)

print("\nNumeric count:", len(numeric_features))
print("Categorical count:", len(categorical_features))
```

```python
from importlib.metadata import version

packages = {
    "streamlit": "streamlit",
    "pandas": "pandas",
    "numpy": "numpy",
    "scipy": "scipy",
    "scikit-learn": "scikit-learn",
    "catboost": "catboost",
    "joblib": "joblib",
}

requirements = [
    f"{requirement_name}=={version(package_name)}"
    for requirement_name, package_name in packages.items()
]

with open("requirements.txt", "w") as file:
    file.write("\n".join(requirements) + "\n")

print("\n".join(requirements))
print("\nSaved requirements.txt")
```
